```python
#Create a Pandas Series containing:
a = [1,2,3,4,5,6,7]
s = pd.Series(a , index = ['i','ii','iii','iv','v','vi','vii'])
print(s)
```

```python
#Create a DataFrame from:
import pandas as pd
data = {
   ' Name': ["Rahul", "Amit", "Priya","gold","silver","bronze","copper"],
'Age': [20, 21, 19,22,23,24,25]
}
d = pd.DataFrame(data)
print(d)
```

```python
#Change the index of a DataFrame.
import pandas as pd
data = {
   ' Name': ["Rahul", "Amit", "Priya","gold","silver","bronze","copper"],
'Age': [20, 21, 19,22,23,24,25]
}
d = pd.DataFrame(data)
d.index = ['i','ii','iii','iv','v','vi','vii']
print(d)
```

```python
#Select rows using loc.
print(d.loc['i'])
```

```python
#Create an empty DataFrame and add three columns to it.
d = pd.DataFrame()
d['name'] = ["Rahul", "Amit", "Priya"]
d['age'] = [20, 21, 19]
d['city'] = ["Delhi", "Mumbai", "Pune"]
print(d)
```
